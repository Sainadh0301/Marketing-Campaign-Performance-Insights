```python
#Plot a scatter plot to visualize the relationship between Acquisition_Cost and ROI.
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.scatterplot(data=df,x='Acquisition_Cost',y='ROI')

plt.title('Acquisition_Cost vs ROI')
plt.xlabel('Acquisition_Cost')
plt.ylabel('ROI')

plt.show()
```

```python
#Create a bar chart to visualize the average Conversion_Rate for different Channel_Used, categorized by Campaign_Type.
plt.figure(figsize=(12,6))
sns.barplot(
    data=df,
    x='Conversion_Rate',
    y='Channel_Used',
    hue='Campaign_Type'
)

plt.title('Average Conversion Rate by Channel and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)

plt.show()
```

```python
#Visualize the distribution of Engagement_Score across different Campaign_Type using a box plot.
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='Campaign_Type',
    y='Engagement_Score'
)
plt.title('Engagement Score by Campaign Type')
plt.xlabel('Campaign Type')
plt.ylabel('Engagement Score')
plt.xticks(rotation=45)

plt.show()
```

```python
#Analyze the average ROI by Company using a bar chart to compare the profitability of campaigns conducted by different companies.
plt.figure(figsize=(10,6))
sns.barplot(
    data=df,
    x='Company',
    y='ROI',
    estimator='mean'
)
plt.title('Average ROI by Company')
plt.xlabel('Company')
plt.ylabel('ROI')
plt.xticks(rotation=45)

plt.show()
```

```python
#Examine the correlation between Engagement_Score and Conversion_Rate using a heatmap.
plt.figure(figsize=(8,5))
corr=df[['Engagement_Score','Conversion_Rate']].corr()
sns.heatmap(corr,annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation between Engagement Score and Conversion Rate')

plt.show()
```

```python
#Create a count plot to visualize the distribution of Target_Audience.
plt.figure(figsize=(10,6))
sns.countplot(
    data=df,
    x='Target_Audience'
)
plt.title('Distribution of Target Audience')
plt.xlabel('Target Audience')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
#Identify which Customer_Segment has the highest Conversion_Rate for each Language using a bar chart.
segment_conversion = df.groupby(
    ['Language', 'Customer_Segment']
)['Conversion_Rate'].mean().reset_index()

idx = segment_conversion.groupby('Language')['Conversion_Rate'].idxmax()

highest_conversion = segment_conversion.loc[idx]

plt.figure(figsize=(12, 6))

sns.barplot(
    data=highest_conversion,
    x='Language',
    y='Conversion_Rate',
    hue='Customer_Segment'
)

plt.title('Highest Conversion Rate by Customer Segment for Each Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)

plt.show()
```

```python
#Visualize the distribution of Acquisition_Cost across each Customer_Segment,categorized by Channel_Used, using a box plot.
plt.figure(figsize=(14, 7))

sns.boxplot(
    data=df,
    x='Customer_Segment',
    y='Acquisition_Cost',
    hue='Channel_Used'
)

plt.title('Acquisition Cost by Customer Segment and Channel')
plt.xlabel('Customer Segment')
plt.ylabel('Acquisition Cost')
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1, 1), loc='upper left')

plt.show()
```

```python
#Analyze average Conversion_Rate by Language using a bar chart to compare the effectiveness of campaigns conducted in different languages.
plt.figure(figsize=(10, 6))

sns.barplot(data=df, x='Language', y='Conversion_Rate')

plt.title('Average Conversion Rate by Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)

plt.show()
```

```python
#Compare the Engagement_Score for different Channels_Used, segmented by Campaign_Type, using a bar chart.
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Channel_Used',
    y='Engagement_Score',
    hue='Campaign_Type'
)

plt.title('Engagement Score by Channel and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Engagement Score')
plt.xticks(rotation=45)
plt.legend(bbox_to_anchor=(1, 1), loc='upper left')

plt.show()
```

```python
#Show the distribution of total ROI across different Channels_Used using a pie chart.
channel_roi = df.groupby('Channel_Used')['ROI'].sum()

plt.figure(figsize=(9, 9))

plt.pie(
    channel_roi,
    labels=channel_roi.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Total ROI by Channel')

plt.show()
```

```python
#Plot a scatter plot to show the relationship between Clicks and Impressions for each Channel_Used.
plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=df,
    x='Impressions',
    y='Clicks',
    hue='Channel_Used'
)

plt.title('Clicks vs Impressions by Channel')
plt.xlabel('Impressions')
plt.ylabel('Clicks')
plt.xticks(rotation=45)

plt.show()
```

```python
#Plot the distribution of Duration using a histogram.
df['Duration_Days'] = df['Duration'].str.replace(' days', '').astype(int)
plt.figure(figsize=(10, 6))

plt.hist(df['Duration_Days'], bins=10, edgecolor='black')

plt.title('Distribution of Campaign Duration')
plt.xlabel('Duration (Days)')
plt.ylabel('Frequency')

plt.show()
```

```python
#Analyze how the overall Conversion_Rate has changed over Date for each Company using a line chart.
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
company_conversion = df.groupby(
    ['Date', 'Company']
)['Conversion_Rate'].mean().reset_index()
plt.figure(figsize=(14, 7))

sns.lineplot(
    data=company_conversion,
    x='Date',
    y='Conversion_Rate',
    hue='Company'
)

plt.title('Conversion Rate Over Time by Company')
plt.xlabel('Date')
plt.ylabel('Average Conversion Rate')

plt.xticks(rotation=45)
plt.legend(title='Company')

plt.show()
```

```python
#Examine the trend of Engagement_Score over Date with a line chart.
daily_engagement = df.groupby('Date')['Engagement_Score'].mean().reset_index()

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=daily_engagement,
    x='Date',
    y='Engagement_Score'
)

plt.title('Engagement Score Over Time')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')

plt.xticks(rotation=45)

plt.show()
```

```python
#Determine which location has the highest Acquisition_Cost using a bar chart.
location_cost = df.groupby('Location')['Acquisition_Cost'].mean().reset_index()

location_cost = location_cost.sort_values(
    by='Acquisition_Cost',
    ascending=False
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=location_cost,
    x='Location',
    y='Acquisition_Cost'
)

plt.title('Average Acquisition Cost by Location')
plt.xlabel('Location')
plt.ylabel('Average Acquisition Cost')
plt.xticks(rotation=45)

plt.show()

highest_location = location_cost.iloc[0]

print("Location with highest Acquisition Cost:",
      highest_location['Location'])
print("Highest Average Acquisition Cost:",
      highest_location['Acquisition_Cost'])
```

```python
#Visualize the Conversion_Rate by different Location, categorized by Target_Audience, using a bar chart.
plt.figure(figsize=(14, 7))

sns.barplot(
    data=df,
    x='Location',
    y='Conversion_Rate',
    hue='Target_Audience'
)

plt.title('Conversion Rate by Location and Target Audience')
plt.xlabel('Location')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```

```python
#Illustrate the proportion of ROI by Location using a pie chart.
location_roi = df.groupby('Location')['ROI'].sum()

plt.figure(figsize=(8, 8))

plt.pie(
    location_roi,
    labels=location_roi.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Proportion of ROI by Location')

plt.show()
```
